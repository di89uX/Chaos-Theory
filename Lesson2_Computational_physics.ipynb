<h1>File Handeling with Pandas</h1>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
u = 0
a = 9.81
t_values = np.linspace(0,5,100) #time from 0 to 5 seconds, 100 points
s_values = u*t_values + 0.5*a*t_values**2

df = pd.DataFrame({'Time (s)': t_values, 'Displacement (m)': s_values})
print(df.head())

df = df.round(4)

df.to_csv('free_fall_displacement_data.csv', index=False)
print("Dataset created and saved!")
```

```python
df = pd.read_csv('/content/free_fall_displacement_data.csv')
#Quick review
print('First 5 rows: ')
print(df.head())
print('Last 5 rows: ')
print(df.tail())

print("\nShape {rows,columns}: ", df.shape)
print("Columns: ", df.columns.tolist())
```

```python
plt.scatter(df['Time (s)'], df['Displacement (m)'], s= 4, label='Free fall', color='blue')
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Free fall displacement')
plt.grid(True)
plt.legend()
plt.show()
```
